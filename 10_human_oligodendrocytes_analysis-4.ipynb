```python
%env PYTHONHASHSEED=0
import os
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib.pyplot import rc_context
import seaborn as sns
import numpy as np
```

```python
OPCa_ipa = pd.read_csv("../output/human/IPA/OPCa_v_OPCcontrol_IPA_pathways_results.txt", sep="\t", header = 1)
OPCa_ipa = OPCa_ipa.sort_values(by=[' -log(p-value)'])
OPCa_ipa = OPCa_ipa[OPCa_ipa['z-score'] > 1]
OPCa_ipa['n_molecules'] = OPCa_ipa['Molecules'].str.count(',')+1
OPCa_ipa = OPCa_ipa[OPCa_ipa['n_molecules'] >= 5]
```

```python
OPCb_ipa = pd.read_csv("../output/human/IPA/OPCb_v_OPCcontrol_IPA_pathways_results.txt", sep="\t", header = 1)
OPCb_ipa = OPCb_ipa.sort_values(by=[' -log(p-value)'])
OPCb_ipa = OPCb_ipa[OPCb_ipa['z-score'] > 1]
OPCb_ipa['n_molecules'] = OPCb_ipa['Molecules'].str.count(',')+1
OPCb_ipa = OPCb_ipa[OPCb_ipa['n_molecules'] >= 5]
```

```python
OPCc_ipa = pd.read_csv("../output/human/IPA/OPCc_v_OPCcontrol_IPA_pathways_results.txt", sep="\t", header = 1)
OPCc_ipa = OPCc_ipa.sort_values(by=[' -log(p-value)'])
OPCc_ipa = OPCc_ipa[OPCc_ipa['z-score'] > 1]
OPCc_ipa['n_molecules'] = OPCc_ipa['Molecules'].str.count(',')+1
OPCc_ipa = OPCc_ipa[OPCc_ipa['n_molecules'] >= 5]
```

```python
OPCd_ipa = pd.read_csv("../output/human/IPA/OPCd_v_OPCcontrol_IPA_pathways_results.txt", sep="\t", header = 1)
OPCd_ipa = OPCd_ipa.sort_values(by=[' -log(p-value)'])
OPCd_ipa = OPCd_ipa[OPCd_ipa['z-score'] > 1]
OPCd_ipa['n_molecules'] = OPCd_ipa['Molecules'].str.count(',')+1
OPCd_ipa = OPCd_ipa[OPCd_ipa['n_molecules'] >= 5]
```

```python
OLGb_ipa = pd.read_csv("../output/human/IPA/OLGb_v_OLGcontrol_IPA_pathways_results.txt", sep="\t", header = 1)
OLGb_ipa = OLGb_ipa.sort_values(by=[' -log(p-value)'])
OLGb_ipa = OLGb_ipa[OLGb_ipa['z-score'] > 1]
OLGb_ipa['n_molecules'] = OLGb_ipa['Molecules'].str.count(',')+1
OLGb_ipa = OLGb_ipa[OLGb_ipa['n_molecules'] >= 5]
```

```python
OLGc_ipa = pd.read_csv("../output/human/IPA/OLGc_v_OLGcontrol_IPA_pathways_results.txt", sep="\t", header = 1)
OLGc_ipa = OLGc_ipa.sort_values(by=[' -log(p-value)'])
OLGc_ipa = OLGc_ipa[OLGc_ipa['z-score'] > 1]
OLGc_ipa['n_molecules'] = OLGc_ipa['Molecules'].str.count(',')+1
OLGc_ipa = OLGc_ipa[OLGc_ipa['n_molecules'] >= 5]
```

```python
OLGd_ipa = pd.read_csv("../output/human/IPA/OLGd_v_OLGcontrol_IPA_pathways_results.txt", sep="\t", header = 1)
OLGd_ipa = OLGd_ipa.sort_values(by=[' -log(p-value)'])
OLGd_ipa = OLGd_ipa[OLGd_ipa['z-score'] > 1]
OLGd_ipa['n_molecules'] = OLGd_ipa['Molecules'].str.count(',')+1
OLGd_ipa = OLGd_ipa[OLGd_ipa['n_molecules'] >= 5]
```

```python
%matplotlib inline
```

```python
# plot OPC-A

color_variable = OPCa_ipa.tail(10)["z-score"]

with rc_context({"figure.figsize": (6, 4), "grid.alpha":0, "figure.dpi":300}):
    # Create the horizontal bar plot
    fig, ax = plt.subplots()
    
    # Normalize color_variable to [0, 1] for the colormap
    norm = plt.Normalize(0, color_variable.max())
    
    # Create a colormap
    cmap = plt.cm.viridis
    
    # Create horizontal bar plot with colors from the colormap
    bars = ax.barh(OPCa_ipa.tail(10)['Ingenuity Canonical Pathways'], OPCa_ipa.tail(10)[' -log(p-value)'], color=cmap(norm(color_variable)))
    
    # Create a colorbar
    sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
    sm.set_array([])  # Only needed for older versions of matplotlib
    cbar = plt.colorbar(sm, ax=ax, anchor=(-0.67, 0), shrink=0.45, aspect = 10, ticks=[1, 2, 3, 4])
    cbar.ax.yaxis.set_ticks_position('left')
    cbar.ax.yaxis.set_label_position('left')
    cbar.set_label('z-score')
    
    # Add labels and title
    ax.set_xlabel('-log10(p-value)')
    ax.set_title('OPC-A versus OPC-Control')
    ax.axvline(x=-1*np.log10(0.05/7), color='black', linestyle='--', alpha = 0.6)
```
