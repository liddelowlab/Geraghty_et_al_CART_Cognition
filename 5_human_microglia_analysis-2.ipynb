```python
import anndata as ad
import numpy as np
import pandas as pd

from scanpro import scanpro, anova, t_test
from scanpro.linear_model import create_design
from scanpro.get_transformed_props import get_transformed_props
```

```python
# load microglia anndata object
adata  = ad.read_h5ad('../output/human/microglia_scanpy_object_intermediate_postclustering.h5ad')
```

```python
adata
```

```python
# test the CART_A sample versus Control samples
out_boot_CART_A = scanpro(adata[adata.obs['sampleid'].isin(["CART_A",
                                 "Control_A", "Control_B", "Control_C"])],
                   clusters_col='choir_clusters_str',
                   conds_col='condition',
                   samples_col = "sampleid",
                   transform='arcsin',
                   )

out_boot_CART_A.results
```

```python
# plot pseudoreplicate simulation proportions & t-test results
out_boot_CART_A.plot()
```

```python
# test CART B
out_boot_CART_B = scanpro(adata[adata.obs['sampleid'].isin(["CART_B",
                                 "Control_A", "Control_B", "Control_C"])],
                   clusters_col='choir_clusters_str',
                   conds_col='condition',
                   samples_col = "sampleid",
                   transform='arcsin'
                   )

out_boot_CART_B.results
```

```python
out_boot_CART_B.plot()
```

```python
# test CART_C
out_boot_CART_C = scanpro(adata[adata.obs['sampleid'].isin(["CART_C",
                                 "Control_A", "Control_B", "Control_C"])],
                   clusters_col='choir_clusters_str',
                   conds_col='condition',
                   samples_col = "sampleid",
                   transform='arcsin'
                   )

out_boot_CART_C.results
```

```python
out_boot_CART_C.plot()
```

```python
# test CART_D
out_boot_CART_D = scanpro(adata[adata.obs['sampleid'].isin(["CART_D",
                                 "Control_A", "Control_B", "Control_C"])],
                   clusters_col='choir_clusters_str',
                   conds_col='condition',
                   samples_col = "sampleid",
                   transform='arcsin'
                   )

out_boot_CART_D.results
```

```python
out_boot_CART_D.plot()
```

```python
cartA_res = out_boot_CART_A.results
cartB_res = out_boot_CART_B.results
cartC_res = out_boot_CART_C.results
cartD_res = out_boot_CART_D.results

cartA_res['Sample'] = 'CAR T A'
cartB_res['Sample'] = 'CAR T B'
cartC_res['Sample'] = 'CAR T C'
cartD_res['Sample'] = 'CAR T D'
```

```python
scanpro_res = pd.concat([cartA_res, cartB_res, cartC_res, cartD_res],
          axis=0, ignore_index = False).reset_index()
```
