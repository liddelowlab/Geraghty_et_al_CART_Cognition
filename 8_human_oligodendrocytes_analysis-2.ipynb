```python
import anndata as ad
import numpy as np
import pandas as pd

from scanpro import scanpro, anova, t_test
from scanpro.linear_model import create_design
from scanpro.get_transformed_props import get_transformed_props
```

```python
# load microglia anndata object
adata  = ad.read_h5ad('../output/human/human_oligodendrocytes_scanpy_object_postclustering.h5ad')
```

```python
adata
```

```python
opcs = adata[adata.obs['final_clusters'].isin(['OPC-Control', 'OPC-A', 'OPC-B', 'OPC-C', 'OPC-D'])]
```

```python
oligs = adata[adata.obs['final_clusters'].isin(['OLG-Control', 'OLG-B', 'OLG-C', 'OLG-D'])]
```

```python
opcs
```

```python
oligs
```

```python
# calculate cell proportions per sample
opc_cell_props = pd.crosstab(opcs.obs['final_clusters'], opcs.obs['sampleid']).apply(lambda col: col / sum(col))
opc_cell_props
```

```python
# Propeller differential abundance testing demonstrated that the OPC-Control cluster significantly decreased in the CAR T samples compared to the Control samples. 
# However, the other clusters in this dataset are largely specific to individual CAR T samples. To evaluate whether these clusters are more abundant in each CAR T sample versus control,
# we now use a linear regression proportion testing approach similar to propeller but which allows single-sample testing by pseudoreplicate bootstrapping from the Scanpro package: https://www.biorxiv.org/content/10.1101/2023.08.14.553234v1
# Note that this approach is not evaluating the biological variance observed between samples, but rather will evaluate whether the proportion of cells observed in given cluster occurs more often than expected based on chance compared to the control condition
```

```python
# test the CART_A sample versus Control samples
out_boot_CART_A_OPC = scanpro(opcs[opcs.obs['sampleid'].isin(["CART_A",
                                 "Control_A", "Control_B", "Control_C"])],
                   clusters_col='final_clusters',
                   conds_col='condition',
                   samples_col = "sampleid",
                   transform='arcsin',
                   )

out_boot_CART_A_OPC.results
```

```python
# plot pseudoreplicate simulation proportions & t-test results
out_boot_CART_A_OPC.plot()
```

```python
# test CART B
out_boot_CART_B_OPC = scanpro(opcs[opcs.obs['sampleid'].isin(["CART_B",
                                 "Control_A", "Control_B", "Control_C"])],
                   clusters_col='final_clusters',
                   conds_col='condition',
                   samples_col = "sampleid",
                   transform='arcsin'
                   )

out_boot_CART_B_OPC.results
```

```python
out_boot_CART_B_OPC.plot()
```

```python
# test CART_C
out_boot_CART_C_OPC = scanpro(opcs[opcs.obs['sampleid'].isin(["CART_C",
                                 "Control_A", "Control_B", "Control_C"])],
                   clusters_col='final_clusters',
                   conds_col='condition',
                   samples_col = "sampleid",
                   transform='arcsin'
                   )

out_boot_CART_C_OPC.results
```

```python
out_boot_CART_C_OPC.plot()
```

```python
# test CART_D
out_boot_CART_D_OPC = scanpro(opcs[opcs.obs['sampleid'].isin(["CART_D",
                                 "Control_A", "Control_B", "Control_C"])],
                   clusters_col='final_clusters',
                   conds_col='condition',
                   samples_col = "sampleid",
                   transform='arcsin'
                   )

out_boot_CART_D_OPC.results
```

```python
out_boot_CART_D_OPC.plot()
```

```python
cartA_res_OPC = out_boot_CART_A_OPC.results
cartB_res_OPC = out_boot_CART_B_OPC.results
cartC_res_OPC = out_boot_CART_C_OPC.results
cartD_res_OPC = out_boot_CART_D_OPC.results

cartA_res_OPC['Sample'] = 'CAR T A'
cartB_res_OPC['Sample'] = 'CAR T B'
cartC_res_OPC['Sample'] = 'CAR T C'
cartD_res_OPC['Sample'] = 'CAR T D'
```

```python
OPC_scanpro_res = pd.concat([cartA_res_OPC, cartB_res_OPC, cartC_res_OPC, cartD_res_OPC],
          axis=0, ignore_index = False).reset_index()
```

```python
OPC_scanpro_res
```

```python
OPC_scanpro_res.to_pickle("../output/human/human_oligos_scanpro_differentialabundance_test_results_OPCs.pkl")
OPC_scanpro_res.to_csv("../output/human/human_oligos_scanpro_differentialabundance_test_results_OPCs.csv")
```
